# Análise de NPS — E-commerce
 
## 1. Entendimento do Negócio
 
Nesta primeira etapa, o objetivo é exercitar o pensamento analítico, não código. As respostas abaixo abordam as perguntas de negócio propostas.
 
### Qual problema de negócio está sendo resolvido?
 
O principal problema é que a empresa trabalha de forma **reativa** — atuando apenas após o cliente já estar insatisfeito — pois não é capaz de entender os fatores que impactam na satisfação do cliente e, consequentemente, no NPS. Apesar de prestar o mesmo serviço a todos os usuários, há uma alta variação na percepção da marca.
 
É necessário que o e-commerce deixe de focar apenas em ações corretivas e institua um processo **proativo** de levantamento de dados operacionais e modelos preditivos, para estudar as causas e prevenir a insatisfação do cliente.
 
### Por que o NPS é importante para um e-commerce?
 
O **Net Promoter Score (NPS)** é uma ferramenta globalmente aceita para medir a satisfação do cliente, além de entregar previsibilidade à empresa quanto a promotores e detratores da marca.
 
Um e-commerce pode manter um rol de clientes a nível global, uma vez que suas operações dependem majoritariamente de transações digitais. Por outro lado, a concorrência também é maior, já que os preços são facilmente comparáveis — portanto, o sucesso da empresa depende ainda mais da satisfação do cliente.
 
### Quais áreas poderiam se beneficiar desses insights?
 
Por ser um indicador consolidado de experiência do cliente, o NPS conecta a experiência a diversas áreas da empresa:
 
| Área | Benefício |
|------|-----------|
| **Logística** | Identificar demandas, sazonalidade, atrasos e possíveis rotas. |
| **Atendimento ao cliente** | Identificar práticas que reduzem a satisfação para evitá-las, ouvir necessidades do cliente para gerar insights e melhorar o tempo de resposta. |
| **Marketing/Comercial** | Identificar possíveis promoções, programas de fidelização e ajustes de valor do produto (satisfação vs. mercado) por meio de estudos de custo-benefício. |
 
### Reflexão: como o NPS impacta...
 
**Recompra**
 
Há uma relação diretamente proporcional entre a satisfação do cliente e a probabilidade de recompra. Um cliente que pontua de 9 a 10 demonstra confiança na empresa, elevando as chances de comprar novamente. A mesma lógica, de forma inversa, aplica-se aos clientes que pontuam de 0 a 6: é improvável que voltem a comprar e não demonstram confiança na empresa.
 
**Boca a boca**
 
Assim como na recompra, o NPS está diretamente ligado à propaganda individual da marca:
 
- **Promotores (9–10):** tendem a enfatizar as qualidades positivas da marca ao recomendá-la para sua rede de contatos.
- **Detratores (0–6):** tendem a diminuir as qualidades da empresa, apontar defeitos para sua rede de contatos e podem recorrer a recursos públicos como o Reclame Aqui ou até a via judicial.
**Market share em e-commerce**
 
No e-commerce, a diferença de preços entre produtos tende a ser baixa, de modo que a experiência do cliente é um diferencial decisivo na escolha do produto ou da marca. Assim, além de ganhar vantagem competitiva, um NPS alto pode abrir margem para expansão de negócios.
 
### Quais indicadores de mercado poderiam complementar essa análise?
 
O NPS isolado mostra apenas a opinião do cliente; combinado a outros indicadores, permite aprofundar nos motivos:
 
| Indicador | O que agrega |
|-----------|--------------|
| **Benchmarks de NPS** | Compara a empresa com as demais do mercado por categoria, ajudando a entender sua posição relativa à concorrência. |
| **SLA logístico** | Levanta dados de entrega (prazos cumpridos, produto e local corretos). |
| **Indicadores de concorrência** | Avalia o que o cliente espera de outras empresas do setor, trazendo melhorias ao processo interno a partir da performance de referência. |
| **Taxa de Churn** | Demonstra a taxa de clientes que abandonaram a empresa em um período; junto ao NPS, permite estabelecer relação causal entre insatisfação e desistência. |
 
Em resumo, quanto mais indicadores combinados, maior a probabilidade de a empresa melhorar seu produto frente à concorrência e oferecer um serviço melhor ao cruzar os resultados de NPS com outros marcadores internos ligados à satisfação do cliente.

## 2. Definição da Target
 
Qual é o alvo desse problema de negócio? Esta etapa avalia o entendimento conceitual, não técnico.
 
### Qual variável representa a satisfação do cliente?
 
A variável que representa a satisfação é a **probabilidade de recomendação**, avaliada pela pergunta *"Em uma escala de 0 a 10, o quanto você recomendaria nossa loja para um amigo ou colega?"*, medida pelo indicador NPS.
 
### Por que ela foi escolhida?
 
A nota de NPS é ideal como target por ser a métrica diretamente conectada ao problema da empresa: reflete a percepção final do cliente e o quanto ele está disposto a agir a favor da marca.
 
O NPS não captura apenas o sentimento sobre uma interação isolada, mas uma percepção mais ampla sobre a relação com a marca. Além disso, a escala de 0 a 10 é padronizada e amplamente utilizada no mercado, facilitando a análise contínua da nota e a categorização dos clientes em **detratores**, **neutros** e **promotores**.
 
### Em que momento da jornada essa informação é coletada?
 
Ao final da jornada de compra, após o encerramento da experiência do cliente com o pedido. Esse é justamente um dos pontos críticos identificados no problema: como a coleta ocorre depois que a experiência já passou, a empresa só identifica clientes insatisfeitos quando o impacto negativo já se concretizou — o que limita a capacidade de agir de forma preventiva.
 
### Existe algum risco de usar essa variável de forma inadequada?
 
O principal risco é tratar o `nps_score` como métrica isolada, sem investigar as causas operacionais por trás da nota. O objetivo não é apenas prever o score, mas entender o que leva a uma nota baixa. Por exemplo, uma combinação de atraso na entrega (`delivery_delay_days`) com múltiplos contatos no atendimento (`customer_service_contacts`) pode indicar um problema logístico que precisa ser corrigido na origem.
 
Outro risco importante é a **natureza subjetiva** da resposta: um cliente pode ter enfrentado diversos problemas durante a jornada e ainda assim atribuir uma nota alta por ter gostado do produto final — e o oposto também é possível. Por isso, o NPS deve ser tratado como o alvo da análise, mas não como a única variável relevante; são as variáveis operacionais que trarão o entendimento acionável para o negócio.

### Quais fatores parecem mais críticos para a satisfação?

Para responder, calculamos a **correlação** de cada variável com o `nps_score`. A correlação mede o quanto duas variáveis "andam juntas".

```python
corr = df.corr(numeric_only=True)

corr['nps_score'].sort_values(ascending=False)
```

Os resultados indicam que a **experiência logística** é o principal fator de satisfação do cliente. **Atrasos na entrega**, **reclamações** e **múltiplos contatos com o atendimento** estão diretamente associados à redução do NPS.

Em contrapartida, clientes que demonstram **fidelização** por meio da recompra tendem a apresentar níveis mais elevados de satisfação.

Dessa forma, ações voltadas à **melhoria da logística** e à **resolução rápida de problemas** podem gerar impactos significativos na percepção dos clientes e no aumento do NPS.

```python

# Criando um dicionário com os fatores que possuem maior correlação com o NPS.
# Valores positivos indicam que o fator tende a aumentar o NPS.
# Valores negativos indicam que o fator tende a reduzir o NPS.

correlacoes = {
    'Atraso na entrega': -0.597,
    'Reclamações': -0.497,
    'Contatos com atendimento': -0.351,
    'Tempo de resolução': -0.191,
    'CSAT': 0.564,
    'Recompra em 30 dias': 0.570
}

# Convertendo o dicionário para um DataFrame.
# Isso facilita a manipulação dos dados e a criação do gráfico.

corr_df = pd.DataFrame(
    correlacoes.items(),
    columns=['Fator', 'Correlação']
)

# Ordenando os dados pela correlação para melhorar a visualização no gráfico.
# Os fatores com maior impacto negativo aparecem primeiro.

corr_df = corr_df.sort_values('Correlação')

# Definindo o tamanho da figura (largura=10, altura=6).

plt.figure(figsize=(10,6))

# Criando um gráfico de barras horizontais.
# No eixo Y ficam os fatores.
# No eixo X ficam os valores de correlação.

plt.barh(corr_df['Fator'], corr_df['Correlação'])

# Adicionando os valores numéricos ao lado de cada barra.
# Isso facilita a interpretação dos resultados.

for i, v in enumerate(corr_df['Correlação']):
    plt.text(v, i, f' {v:.3f}')

# Adicionando título ao gráfico.

plt.title('Fatores que mais impactam o NPS')

# Nomeando os eixos.

plt.xlabel('Correlação com o NPS')
plt.ylabel('Fatores')

# Adicionando linhas de grade para facilitar a leitura dos valores.

plt.grid(axis='x', linestyle='--', alpha=0.5)

# Exibindo o gráfico.

plt.show()
```

**Explicando esse gráfico**

O gráfico acima mostra a **correlação** de cada fator com o NPS. De forma simples:

- Valores próximos de **+1** indicam uma relação positiva forte — quando uma variável aumenta, a outra também tende a aumentar.
- Valores próximos de **-1** indicam uma relação negativa forte — quando uma variável aumenta, a outra tende a diminuir.
- Valores próximos de **0** indicam pouca ou nenhuma relação entre as variáveis.

Na análise realizada, a variável `repeat_purchase_30d` (recompra em até 30 dias após o pedido) apresentou correlação de **0,570** com o NPS. Isso significa que clientes que realizam novas compras em um curto período tendem a apresentar níveis mais elevados de satisfação, refletidos em notas maiores de NPS.

Por outro lado, a variável `delivery_delay_days` (dias de atraso na entrega) apresentou correlação de **-0,597**, a mais forte entre os fatores analisados. Esse resultado indica que quanto maior o atraso na entrega, menor tende a ser a nota atribuída pelo cliente, evidenciando que a logística é um dos principais fatores que impactam negativamente a experiência do consumidor.

Além disso, variáveis como `complaints_count` (quantidade de reclamações) e `customer_service_contacts` (quantidade de contatos com o atendimento) também apresentaram correlações negativas relevantes, demonstrando que clientes que enfrentam problemas e precisam acionar o suporte tendem a avaliar a empresa de forma menos favorável.

Dessa forma, a análise sugere que os principais fatores relacionados à satisfação dos clientes estão associados à **eficiência logística**, à **qualidade do atendimento** e à capacidade da empresa de proporcionar uma experiência positiva que incentive novas compras.

### O que mais gera detratores?

Olhamos novamente para as correlações negativas com o NPS — os fatores que "puxam a nota para baixo".

A análise indica que o principal gerador de detratores é o **atraso na entrega**, que apresentou a correlação negativa mais forte com o NPS (**-0,597**).

Além disso, clientes que registraram mais **reclamações** (**-0,497**) e precisaram realizar **múltiplos contatos com o atendimento** (**-0,351**) também tendem a avaliar a empresa de forma mais negativa.

Outro fator relevante é o **tempo de resolução de problemas** (**-0,191**), sugerindo que processos lentos de suporte contribuem para a insatisfação dos clientes.

Dessa forma, os resultados indicam que questões relacionadas à **logística** e ao **atendimento** são os principais fatores responsáveis pela geração de detratores.

```python
detratores = {
    'Atraso na entrega': -0.597,
    'Reclamações': -0.497,
    'Contatos com atendimento': -0.351,
    'Tempo de resolução': -0.191
}

detratores_df = pd.DataFrame(
    detratores.items(),
    columns=['Fator', 'Correlação']
)

detratores_df = detratores_df.sort_values('Correlação')

plt.figure(figsize=(10,5))
plt.barh(detratores_df['Fator'], detratores_df['Correlação'])

for i, v in enumerate(detratores_df['Correlação']):
    plt.text(v, i, f' {v:.3f}')

plt.title('Principais Geradores de Detratores')
plt.xlabel('Correlação com o NPS')
plt.ylabel('Fatores')
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.show()
```

### Existe algum "ponto de ruptura" na experiência do cliente?

Para investigar, observamos o **NPS médio em função dos dias de atraso na entrega**.

```python
df.groupby('delivery_delay_days')['nps_score'].mean().round(2)
```

A análise do **NPS médio por dias de atraso** indica a existência de um **ponto de ruptura** na experiência do cliente.

- Sem atraso, o NPS médio é de **6,86 pontos**.
- Com apenas **1 dia** de atraso, o indicador já apresenta uma queda significativa para **5,55 pontos**.
- A partir de **2 dias** de atraso, o NPS passa para **4,58 pontos**, indicando um aumento expressivo na quantidade de detratores.

Os resultados sugerem que **atrasos superiores a 2 dias** representam um ponto crítico na experiência do cliente, provocando uma redução acentuada da satisfação e impactando negativamente a percepção sobre a empresa.

```python
atraso = df.groupby('delivery_delay_days')['nps_score'].mean().round(2)

plt.figure(figsize=(10,5))
plt.plot(atraso.index, atraso.values, marker='o')

for x, y in zip(atraso.index, atraso.values):
    plt.text(x, y, f'{y}')

plt.title('Impacto do atraso na entrega sobre o NPS')
plt.xlabel('Dias de atraso')
plt.ylabel('NPS médio')
plt.grid(True)

plt.show()
```

### Que tipo de cliente tende a ter NPS mais alto ou mais baixo?

Comparamos o **NPS médio** entre clientes que realizaram recompra em até 30 dias e clientes que não recompraram.

Os clientes que realizaram **recompra em até 30 dias** após o pedido tendem a apresentar os maiores níveis de satisfação, com **NPS médio de 9,01 pontos**. Por outro lado, clientes que **não** realizaram uma nova compra apresentam **NPS médio de apenas 3,94 pontos**.

Esse resultado sugere uma forte relação entre satisfação e fidelização: experiências positivas aumentam significativamente a probabilidade de recompra. Na prática, o cliente "ideal" — com NPS mais alto — é aquele que recebeu o pedido **no prazo**, **não precisou reclamar** nem acionar o atendimento e, por isso, **voltou a comprar**. Já o cliente com NPS mais baixo é o que enfrentou **atrasos**, **reclamações** e **múltiplos contatos** com o suporte.

```python
df.groupby('repeat_purchase_30d')['nps_score'].mean()
```

```python
nps_recompra = df.groupby('repeat_purchase_30d')['nps_score'].mean()

plt.figure(figsize=(8,5))
plt.bar(['Não recomprou', 'Recomprou'], nps_recompra)

for i, v in enumerate(nps_recompra):
    plt.text(i, v + 0.1, f'{v:.2f}', ha='center')

plt.title('NPS Médio por Recompra em 30 Dias')
plt.xlabel('Cliente realizou recompra?')
plt.ylabel('NPS Médio')
plt.ylim(0, 10)

plt.show()
```
